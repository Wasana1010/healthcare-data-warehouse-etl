#### Clean invalid waiting times and missing fees.
#### Expected output: The number of changed rows for each rule.

```python
invalid_wait = appointments["wait_minutes"] < 0
wait_changed = invalid_wait.sum()

appointments.loc[invalid_wait, "wait_minutes"] = pd.NA

median_fee = appointments["fee"].median()
missing_fee = appointments["fee"].isnull()
fee_changed = missing_fee.sum()

appointments.loc[missing_fee, "fee"] = median_fee

print("wait_minutes changed:", wait_changed)
print("fee changed:", fee_changed)
print("Median fee used:", median_fee)
```

#### Create surrogate keys for the dimension tables.
#### Expected output: Each dimension row has a unique surrogate key.

```python
dim_patient = patients.copy()
dim_doctor = doctors.copy()
dim_clinic = clinics.copy()

dim_patient.insert(0, "patient_key", range(1, len(dim_patient) + 1))
dim_doctor.insert(0, "doctor_key", range(1, len(dim_doctor) + 1))
dim_clinic.insert(0, "clinic_key", range(1, len(dim_clinic) + 1))

print("Patients:", len(dim_patient))
print("Doctors:", len(dim_doctor))
print("Clinics:", len(dim_clinic))
```

#### Replace source IDs in the fact data with surrogate keys.
#### Expected output: The fact data contains the four surrogate keys.

```python
fact = appointments.copy()

fact = fact.merge(
    dim_patient[["patient_id", "patient_key"]],
    on="patient_id",
    how="left"
)

fact = fact.merge(
    dim_doctor[["doctor_id", "doctor_key"]],
    on="doctor_id",
    how="left"
)

fact = fact.merge(
    dim_clinic[["clinic_id", "clinic_key"]],
    on="clinic_id",
    how="left"
)

print(fact[[
    "patient_key",
    "doctor_key",
    "clinic_key"
]].head())
```

#### Create the 2025 date dimension.
#### Expected output: One row for every date in 2025.

```python
dates = pd.date_range(
    start="2025-01-01",
    end="2025-12-31",
    freq="D"
)

dim_date = pd.DataFrame({
    "date_key": range(1, len(dates) + 1),
    "full_date": dates,
    "year": dates.year,
    "quarter": dates.quarter,
    "month": dates.month,
    "day_name": dates.day_name()
})

print("Date rows:", len(dim_date))
print(dim_date.head())
```

#### Add the date surrogate key to the fact data.
#### Expected output: Each appointment has a date_key.

```python
fact = fact.merge(
    dim_date[["full_date", "date_key"]],
    left_on="appointment_date",
    right_on="full_date",
    how="left"
)

print(fact[[
    "appointment_date",
    "date_key"
]].head())

print("Missing date keys:", fact["date_key"].isnull().sum())
```

#### Prepare the final fact table with surrogate keys and measures.
#### Expected output: The fact table contains four foreign keys and three measures.

```python
fact_appointment = fact[[
    "patient_key",
    "doctor_key",
    "clinic_key",
    "date_key",
    "wait_minutes",
    "consultation_minutes",
    "fee"
]].copy()

print(fact_appointment.shape)
print(fact_appointment.columns.tolist())
```

#### Prepare the final fact table

```python
fact_appointment["wait_minutes"] = fact_appointment["wait_minutes"].astype("Int64")
fact_appointment["consultation_minutes"] = fact_appointment["consultation_minutes"].astype("Int64")
```

#### Export the five final warehouse tables as CSV files.
#### Expected output: Five CSV files are created for the warehouse load.

```python
import os

output_folder = r"D:\DW cw data\cw-c-data"

os.makedirs(output_folder, exist_ok=True)

dim_patient.to_csv(
    os.path.join(output_folder, "dim_patient.csv"),
    index=False
)

dim_doctor.to_csv(
    os.path.join(output_folder, "dim_doctor.csv"),
    index=False
)

dim_clinic.to_csv(
    os.path.join(output_folder, "dim_clinic.csv"),
    index=False
)

dim_date.to_csv(
    os.path.join(output_folder, "dim_date.csv"),
    index=False
)

fact_appointment.to_csv(
    os.path.join(output_folder, "fact_appointment.csv"),
    index=False
)

print("Five CSV files exported successfully.")
```
